```python
import pandas as pd
import random
from datetime import datetime, timedelta

# Set seed for reproducibility
random.seed(42)

# Configuration
num_records = 25000

# Data components
channels = ['Email', 'Phone', 'Chat', 'Social Media', 'In-Store', 'email', 'PHONE', 'Chat ']
categories = ['Laptops', 'Phones', 'Accessories', 'Smart Home', 'Other', 'laptop', '']
products = [
    'iPhone 14', 'iphone14', 'Samsung Galaxy S23', 'MacBook Pro', 'macbook pro',
    'Dell XPS 15', 'AirPods Pro', 'airpods', 'Sony Headphones', 'Ring Doorbell',
    'Nest Thermostat', 'Logitech Mouse', 'HP Laptop', '', 'Unknown'
]
stores = ['Store #1', 'Store #2', 'Store #3', 'Store #7', 'Store #12', 
          'Store #15', 'Store #unknown', 'Online', '', 'store 7']
priorities = ['Low', 'Medium', 'High', 'Critical', '', 'low', 'HIGH']
statuses = ['Resolved', 'Pending', 'Escalated', 'Closed-Unresolved']
complaint_keywords = [
    'product broken on arrival', 'delivery delayed by 2 weeks', 'rude customer service',
    'wrong item shipped', 'refund not processed', 'technical issue not working',
    'defective product', 'poor quality', 'overcharged on bill', 'website crash during checkout',
    'long wait time', 'unprofessional staff', 'damaged packaging', 'missing accessories'
]

# Generate data
data = []
start_date = datetime(2022, 6, 1)

for i in range(num_records):
    complaint_id = f"C{10000 + i}"
    customer_id = f"CUST{random.randint(1000, 8000)}"
    
    # Random date with different formats
    days_offset = random.randint(0, 730)
    complaint_date = start_date + timedelta(days=days_offset)
    date_formats = [
        complaint_date.strftime('%Y-%m-%d'),
        complaint_date.strftime('%m/%d/%y'),
        complaint_date.strftime('%B %d, %Y'),
        complaint_date.strftime('%d-%m-%Y')
    ]
    date_str = random.choice(date_formats)
    
    channel = random.choice(channels)
    category = random.choice(categories)
    product = random.choice(products)
    store = random.choice(stores) if random.random() > 0.3 else ''
    
    # Complaint text
    complaint_text = random.choice(complaint_keywords) + ". " + random.choice([
        "Very disappointed with service", "Need immediate resolution", 
        "This is unacceptable", "Please help", "Frustrated customer"
    ])
    
    priority = random.choice(priorities)
    
    # Resolution time with outliers
    if random.random() > 0.95:
        resolution_hours = random.choice([0, -5, 5000, 9999])  # Outliers
    else:
        resolution_hours = random.randint(1, 200)
    
    status = random.choice(statuses)
    
    # Satisfaction (30% missing)
    satisfaction = random.randint(1, 5) if random.random() > 0.3 else None
    
    repeat_complaint = random.choice(['Yes', 'No', 'YES', 'no', ''])
    customer_tenure = random.randint(1, 60)
    lifetime_value = round(random.uniform(100, 5000), 2) if random.random() > 0.1 else None
    
    # Add duplicate records occasionally
    if random.random() > 0.98:
        data.append([complaint_id, customer_id, date_str, channel, category, product, 
                    store, complaint_text, priority, resolution_hours, status, 
                    satisfaction, repeat_complaint, customer_tenure, lifetime_value])
    
    data.append([complaint_id, customer_id, date_str, channel, category, product, 
                store, complaint_text, priority, resolution_hours, status, 
                satisfaction, repeat_complaint, customer_tenure, lifetime_value])

# Create DataFrame
columns = ['complaint_id', 'customer_id', 'date_submitted', 'channel', 'product_category',
           'product_name', 'store_location', 'complaint_text', 'priority_level',
           'resolution_time_hours', 'resolution_status', 'satisfaction_score',
           'repeat_complaint', 'customer_tenure_months', 'lifetime_value']

df = pd.DataFrame(data, columns=columns)

# Save to CSV
df.to_csv('customer_complaints_raw.csv', index=False)
print(f"Dataset created: {len(df)} records")
print(f"File saved as: customer_complaints_raw.csv")
print("\nFirst few rows:")
print(df.head())
```
