```python
# Day 6: Extract Key Business Insights
# Goal: Summarize findings for final report

import pandas as pd

print("="*60)
print("EXTRACTING KEY BUSINESS INSIGHTS")
print("="*60)

# Load all results
df_main = pd.read_csv('customer_complaints_enriched.csv')
df_categories = pd.read_csv('sql_results/q2_complaint_categories.csv')
df_resolution = pd.read_csv('sql_results/q3_resolution_by_category.csv')
df_channels = pd.read_csv('sql_results/q4_channel_performance.csv')
df_products = pd.read_csv('sql_results/q5_top_products.csv')
df_stores = pd.read_csv('sql_results/q6_store_performance.csv')
df_segments = pd.read_csv('sql_results/q7_customer_segment_comparison.csv')
df_status = pd.read_csv('sql_results/q9_resolution_status.csv')

print("\n" + "="*60)
print("INSIGHT 1: COMPLAINT VOLUME & TRENDS")
print("="*60)
total = len(df_main)
resolved = df_status[df_status['resolution_status']=='resolved']['total_complaints'].sum()
resolved_pct = (resolved / total) * 100
print(f"• Total complaints analyzed: {total:,}")
print(f"• Successfully resolved: {resolved:,} ({resolved_pct:.1f}%)")
print(f"• Average resolution time: {df_main['resolution_time_hours'].mean():.1f} hours")
print(f"• Average satisfaction score: {df_main['satisfaction_score'].mean():.2f}/5.0")

print("\n" + "="*60)
print("INSIGHT 2: TOP PROBLEM AREAS")
print("="*60)
print("Top 3 complaint categories:")
for i, row in df_categories.head(3).iterrows():
    print(f"  {i+1}. {row['complaint_category'].title()}: {row['total_complaints']} ({row['percentage']}%)")

print("\nSlowest to resolve:")
slowest = df_resolution.nlargest(3, 'avg_resolution_hours')
for i, row in slowest.iterrows():
    print(f"  • {row['complaint_category'].title()}: {row['avg_resolution_hours']:.1f} hours")

print("\n" + "="*60)
print("INSIGHT 3: CHANNEL PERFORMANCE")
print("="*60)
fastest_channel = df_channels.nsmallest(1, 'avg_resolution_hours').iloc[0]
slowest_channel = df_channels.nlargest(1, 'avg_resolution_hours').iloc[0]
print(f"• Fastest channel: {fastest_channel['channel'].title()} ({fastest_channel['avg_resolution_hours']:.1f}h)")
print(f"• Slowest channel: {slowest_channel['channel'].title()} ({slowest_channel['avg_resolution_hours']:.1f}h)")
print(f"• Difference: {slowest_channel['avg_resolution_hours'] - fastest_channel['avg_resolution_hours']:.1f} hours")

print("\n" + "="*60)
print("INSIGHT 4: PRODUCT QUALITY ISSUES")
print("="*60)
print("Top 3 most complained products:")
for i, row in df_products.head(3).iterrows():
    sat = row['avg_satisfaction'] if pd.notna(row['avg_satisfaction']) else 0
    print(f"  {i+1}. {row['product_name'].title()}: {row['complaint_count']} complaints (★{sat:.1f})")

print("\n" + "="*60)
print("INSIGHT 5: STORE PERFORMANCE")
print("="*60)
best_store = df_stores.nlargest(1, 'avg_satisfaction').iloc[0]
worst_store = df_stores.nsmallest(1, 'avg_satisfaction').iloc[0]
print(f"• Best performing: {best_store['store_location']} (★{best_store['avg_satisfaction']:.2f})")
print(f"• Needs improvement: {worst_store['store_location']} (★{worst_store['avg_satisfaction']:.2f})")

busiest_store = df_stores.nlargest(1, 'total_complaints').iloc[0]
print(f"• Most complaints: {busiest_store['store_location']} ({busiest_store['total_complaints']} complaints)")

print("\n" + "="*60)
print("INSIGHT 6: CUSTOMER SEGMENT ANALYSIS")
print("="*60)
high_value = df_segments[df_segments['is_high_value_customer']=='High-Value'].iloc[0]
regular = df_segments[df_segments['is_high_value_customer']=='Regular'].iloc[0]
print(f"• High-value customers:")
print(f"  - Resolution time: {high_value['avg_resolution_hours']:.1f}h")
print(f"  - Satisfaction: ★{high_value['avg_satisfaction']:.2f}")
print(f"  - Fast resolution rate: {high_value['pct_fast_resolution']:.1f}%")
print(f"\n• Regular customers:")
print(f"  - Resolution time: {regular['avg_resolution_hours']:.1f}h")
print(f"  - Satisfaction: ★{regular['avg_satisfaction']:.2f}")
print(f"  - Fast resolution rate: {regular['pct_fast_resolution']:.1f}%")

time_diff = abs(high_value['avg_resolution_hours'] - regular['avg_resolution_hours'])
if high_value['avg_resolution_hours'] < regular['avg_resolution_hours']:
    print(f"\n⚠ High-value customers get {time_diff:.1f}h faster service")
else:
    print(f"\n⚠ Regular customers get {time_diff:.1f}h faster service")

print("\n" + "="*60)
print("INSIGHT 7: REPEAT COMPLAINTS")
print("="*60)
repeat_count = df_main['is_repeat_complaint'].sum()
repeat_pct = (repeat_count / len(df_main)) * 100
print(f"• Repeat complaints: {repeat_count:,} ({repeat_pct:.1f}%)")
print(f"• This suggests {repeat_pct:.1f}% of issues weren't properly resolved initially")

print("\n" + "="*60)
print("INSIGHT 8: TIME PATTERNS")
print("="*60)
weekend_complaints = df_main['is_weekend'].sum()
weekend_pct = (weekend_complaints / len(df_main)) * 100
holiday_complaints = df_main['is_holiday_season'].sum()
holiday_pct = (holiday_complaints / len(df_main)) * 100
print(f"• Weekend complaints: {weekend_complaints:,} ({weekend_pct:.1f}%)")
print(f"• Holiday season complaints: {holiday_complaints:,} ({holiday_pct:.1f}%)")

print("\n" + "="*60)
print("✅ INSIGHTS EXTRACTION COMPLETE")
print("="*60)
print("\nThese insights will be used in Day 7 final report!")
```
