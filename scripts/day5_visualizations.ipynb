```python
# Day 5: Python Visualizations (Part 1)
# Goal: Create insightful visualizations from SQL results

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set style for better-looking charts
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

# Create folder for saving charts
import os
os.makedirs('visualizations', exist_ok=True)

print("="*60)
print("DAY 5: CREATING VISUALIZATIONS")
print("="*60)

# ============================================================
# VISUALIZATION 1: Monthly Complaint Trends (Line Chart)
# ============================================================
print("\n📊 Creating Chart 1: Monthly Complaint Trends")

df1 = pd.read_csv('sql_results/q1_monthly_trends.csv')

# Create date column for better plotting
df1['date'] = pd.to_datetime(df1['year'].astype(str) + '-' + df1['month'].astype(str) + '-01')

plt.figure(figsize=(14, 6))
plt.plot(df1['date'], df1['total_complaints'], marker='o', linewidth=2, markersize=6, color='#2E86AB')
plt.title('Customer Complaints Over Time', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Number of Complaints', fontsize=12)
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('visualizations/01_monthly_trends.png', dpi=300, bbox_inches='tight')
plt.close()
print("✅ Saved: visualizations/01_monthly_trends.png")

# ============================================================
# VISUALIZATION 2: Complaint Categories (Horizontal Bar)
# ============================================================
print("\n📊 Creating Chart 2: Complaint Categories")

df2 = pd.read_csv('sql_results/q2_complaint_categories.csv')
df2 = df2.sort_values('total_complaints')

plt.figure(figsize=(10, 6))
colors = sns.color_palette("Blues_r", len(df2))
plt.barh(df2['complaint_category'], df2['total_complaints'], color=colors)
plt.title('Complaints by Category', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Number of Complaints', fontsize=12)
plt.ylabel('Category', fontsize=12)

# Add value labels on bars
for i, v in enumerate(df2['total_complaints']):
    plt.text(v + 50, i, f"{v} ({df2['percentage'].iloc[i]}%)", va='center', fontsize=10)

plt.tight_layout()
plt.savefig('visualizations/02_complaint_categories.png', dpi=300, bbox_inches='tight')
plt.close()
print("✅ Saved: visualizations/02_complaint_categories.png")

# ============================================================
# VISUALIZATION 3: Resolution Time by Category (Bar Chart)
# ============================================================
print("\n📊 Creating Chart 3: Resolution Time by Category")

df3 = pd.read_csv('sql_results/q3_resolution_by_category.csv')
df3 = df3.sort_values('avg_resolution_hours', ascending=False)

fig, ax = plt.subplots(figsize=(12, 6))
bars = ax.bar(df3['complaint_category'], df3['avg_resolution_hours'], 
              color=['#E63946' if x > 60 else '#2E86AB' for x in df3['avg_resolution_hours']])
plt.title('Average Resolution Time by Complaint Category', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Category', fontsize=12)
plt.ylabel('Average Hours to Resolve', fontsize=12)
plt.xticks(rotation=45, ha='right')

# Add horizontal line for target (48 hours)
plt.axhline(y=48, color='red', linestyle='--', linewidth=2, label='Target: 48 hours', alpha=0.7)
plt.legend()

# Add value labels
for i, v in enumerate(df3['avg_resolution_hours']):
    ax.text(i, v + 2, f'{v:.1f}h', ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.savefig('visualizations/03_resolution_by_category.png', dpi=300, bbox_inches='tight')
plt.close()
print("✅ Saved: visualizations/03_resolution_by_category.png")

# ============================================================
# VISUALIZATION 4: Channel Performance Comparison
# ============================================================
print("\n📊 Creating Chart 4: Channel Performance")

df4 = pd.read_csv('sql_results/q4_channel_performance.csv')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Chart 4a: Resolution time by channel
ax1.bar(df4['channel'], df4['avg_resolution_hours'], color='#06A77D')
ax1.set_title('Avg Resolution Time by Channel', fontsize=14, fontweight='bold')
ax1.set_xlabel('Channel', fontsize=11)
ax1.set_ylabel('Hours', fontsize=11)
ax1.tick_params(axis='x', rotation=45)

for i, v in enumerate(df4['avg_resolution_hours']):
    ax1.text(i, v + 1, f'{v:.1f}h', ha='center', va='bottom', fontsize=9)

# Chart 4b: Satisfaction by channel
ax2.bar(df4['channel'], df4['avg_satisfaction'], color='#F77F00')
ax2.set_title('Avg Satisfaction by Channel', fontsize=14, fontweight='bold')
ax2.set_xlabel('Channel', fontsize=11)
ax2.set_ylabel('Satisfaction Score (1-5)', fontsize=11)
ax2.set_ylim(0, 5)
ax2.tick_params(axis='x', rotation=45)

for i, v in enumerate(df4['avg_satisfaction']):
    if pd.notna(v):
        ax2.text(i, v + 0.1, f'{v:.2f}', ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.savefig('visualizations/04_channel_performance.png', dpi=300, bbox_inches='tight')
plt.close()
print("✅ Saved: visualizations/04_channel_performance.png")

# ============================================================
# VISUALIZATION 5: Top 10 Problem Products
# ============================================================
print("\n📊 Creating Chart 5: Top Problem Products")

df5 = pd.read_csv('sql_results/q5_top_products.csv')

fig, ax = plt.subplots(figsize=(12, 8))
bars = ax.barh(df5['product_name'], df5['complaint_count'], color='#D62828')
ax.set_title('Top 10 Most Complained Products', fontsize=16, fontweight='bold', pad=20)
ax.set_xlabel('Number of Complaints', fontsize=12)
ax.set_ylabel('Product', fontsize=12)

# Add value labels
for i, v in enumerate(df5['complaint_count']):
    ax.text(v + 5, i, f"{v}", va='center', fontsize=10)

# Add satisfaction scores as text
for i, (count, sat) in enumerate(zip(df5['complaint_count'], df5['avg_satisfaction'])):
    if pd.notna(sat):
        ax.text(count + 50, i, f"★{sat:.1f}", va='center', fontsize=9, color='orange')

plt.tight_layout()
plt.savefig('visualizations/05_top_products.png', dpi=300, bbox_inches='tight')
plt.close()
print("✅ Saved: visualizations/05_top_products.png")

# ============================================================
# VISUALIZATION 6: Store Performance Matrix
# ============================================================
print("\n📊 Creating Chart 6: Store Performance")

df6 = pd.read_csv('sql_results/q6_store_performance.csv')

fig, ax = plt.subplots(figsize=(10, 6))
scatter = ax.scatter(df6['avg_resolution_hours'], df6['avg_satisfaction'], 
                     s=df6['total_complaints']*3, alpha=0.6, c=df6['total_complaints'],
                     cmap='YlOrRd', edgecolors='black', linewidth=1)

# Add store labels
for i, store in enumerate(df6['store_location']):
    ax.annotate(store, (df6['avg_resolution_hours'].iloc[i], df6['avg_satisfaction'].iloc[i]),
                fontsize=9, ha='center')

ax.set_title('Store Performance: Resolution Time vs Satisfaction', fontsize=16, fontweight='bold', pad=20)
ax.set_xlabel('Avg Resolution Time (hours)', fontsize=12)
ax.set_ylabel('Avg Satisfaction Score', fontsize=12)
ax.axhline(y=3.5, color='red', linestyle='--', alpha=0.5, label='Target Satisfaction: 3.5')
ax.axvline(x=50, color='blue', linestyle='--', alpha=0.5, label='Target Resolution: 50h')
ax.legend()

# Add colorbar
cbar = plt.colorbar(scatter, ax=ax)
cbar.set_label('Total Complaints', fontsize=11)

plt.tight_layout()
plt.savefig('visualizations/06_store_performance.png', dpi=300, bbox_inches='tight')
plt.close()
print("✅ Saved: visualizations/06_store_performance.png")

# ============================================================
# VISUALIZATION 7: Customer Segment Comparison
# ============================================================
print("\n📊 Creating Chart 7: Customer Segment Comparison")

df7 = pd.read_csv('sql_results/q7_customer_segment_comparison.csv')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Resolution time comparison
x = range(len(df7))
width = 0.35
ax1.bar([i - width/2 for i in x], df7['avg_resolution_hours'], width, 
        label='Resolution Time', color='#06A77D')
ax1.bar([i + width/2 for i in x], df7['avg_satisfaction']*10, width,
        label='Satisfaction (x10)', color='#F77F00')
ax1.set_title('High-Value vs Regular Customers', fontsize=14, fontweight='bold')
ax1.set_ylabel('Hours / Score', fontsize=11)
ax1.set_xticks(x)
ax1.set_xticklabels(df7['is_high_value_customer'])
ax1.legend()

# Add values on bars
for i, (res, sat) in enumerate(zip(df7['avg_resolution_hours'], df7['avg_satisfaction'])):
    ax1.text(i - width/2, res + 1, f'{res:.1f}h', ha='center', va='bottom', fontsize=9)
    ax1.text(i + width/2, sat*10 + 0.5, f'{sat:.2f}', ha='center', va='bottom', fontsize=9)

# Fast resolution percentage
ax2.bar(df7['is_high_value_customer'], df7['pct_fast_resolution'], color='#2E86AB')
ax2.set_title('% of Complaints Resolved Quickly (<24h)', fontsize=14, fontweight='bold')
ax2.set_ylabel('Percentage (%)', fontsize=11)

for i, v in enumerate(df7['pct_fast_resolution']):
    ax2.text(i, v + 1, f'{v:.1f}%', ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.savefig('visualizations/07_customer_segments.png', dpi=300, bbox_inches='tight')
plt.close()
print("✅ Saved: visualizations/07_customer_segments.png")

# ============================================================
# VISUALIZATION 8: Day of Week Patterns
# ============================================================
print("\n📊 Creating Chart 8: Day of Week Patterns")

df10 = pd.read_csv('sql_results/q10_day_patterns.csv')

# Sort by day order
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
df10['day_of_week'] = pd.Categorical(df10['day_of_week'], categories=day_order, ordered=True)
df10 = df10.sort_values('day_of_week')

plt.figure(figsize=(12, 6))
colors = ['#E63946' if weekend else '#2E86AB' for weekend in df10['is_weekend']]
bars = plt.bar(df10['day_of_week'], df10['total_complaints'], color=colors)

plt.title('Complaints by Day of Week', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Day', fontsize=12)
plt.ylabel('Number of Complaints', fontsize=12)

# Add value labels
for i, v in enumerate(df10['total_complaints']):
    plt.text(i, v + 50, str(v), ha='center', va='bottom', fontsize=10)

# Add legend
from matplotlib.patches import Patch
legend_elements = [Patch(facecolor='#2E86AB', label='Weekday'),
                   Patch(facecolor='#E63946', label='Weekend')]
plt.legend(handles=legend_elements)

plt.tight_layout()
plt.savefig('visualizations/08_day_patterns.png', dpi=300, bbox_inches='tight')
plt.close()
print("✅ Saved: visualizations/08_day_patterns.png")

print("\n" + "="*60)
print("✅ DAY 5 PART 1 COMPLETE!")
print("="*60)
print("\nCreated 8 visualizations in 'visualizations/' folder")
print("\nKey Insights So Far:")
print("1. Monthly trends show complaint patterns")
print("2. Identified top complaint categories")
print("3. Resolution times vary significantly by type")
print("4. Some channels perform better than others")
print("5. Specific products need attention")
print("6. Store performance varies widely")
print("7. High-value customers may get different service")
print("8. Weekday vs weekend patterns exist")
```
