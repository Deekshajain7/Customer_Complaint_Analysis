```python
# Day 1: Initial Data Exploration
# Goal: Understand our raw data before cleaning

import pandas as pd

# Load the data
df = pd.read_csv('customer_complaints_raw.csv')

print("="*60)
print("STEP 1: BASIC DATASET INFO")
print("="*60)

# How many rows and columns?
print(f"\nDataset Shape: {df.shape[0]} rows, {df.shape[1]} columns")

# What columns do we have?
print("\nColumn Names:")
print(df.columns.tolist())

print("\n" + "="*60)
print("STEP 2: FIRST LOOK AT THE DATA")
print("="*60)

# See first 5 rows
print("\nFirst 5 rows:")
print(df.head())

print("\n" + "="*60)
print("STEP 3: DATA TYPES")
print("="*60)

# What type is each column? (number, text, etc.)
print("\nData Types:")
print(df.dtypes)

print("\n" + "="*60)
print("STEP 4: MISSING VALUES")
print("="*60)

# Count missing values in each column
missing = df.isnull().sum()
missing_percent = (missing / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing_Count': missing,
    'Missing_Percent': missing_percent
})
print("\nMissing Values:")
print(missing_df[missing_df['Missing_Count'] > 0])

print("\n" + "="*60)
print("STEP 5: BASIC STATISTICS (Numbers Only)")
print("="*60)

# Statistics for numeric columns
print(df.describe())

print("\n" + "="*60)
print("STEP 6: UNIQUE VALUES IN KEY COLUMNS")
print("="*60)

# Check unique values in important columns
columns_to_check = ['channel', 'product_category', 'priority_level', 'resolution_status']

for col in columns_to_check:
    print(f"\n{col}: {df[col].nunique()} unique values")
    print(df[col].value_counts())

print("\n" + "="*60)
print("STEP 7: DUPLICATE COMPLAINT IDs")
print("="*60)

# Check for duplicate complaints
duplicates = df[df.duplicated(subset=['complaint_id'], keep=False)]
print(f"\nTotal duplicate complaint IDs: {len(duplicates)}")
print(f"Unique complaint IDs with duplicates: {duplicates['complaint_id'].nunique()}")

print("\n" + "="*60)
print("DAY 1 EXPLORATION COMPLETE!")
print("="*60)
print("\nKEY FINDINGS:")
print(f"1. Total records: {len(df)}")
print(f"2. Missing satisfaction scores: {missing_df.loc['satisfaction_score', 'Missing_Count']:.0f} ({missing_df.loc['satisfaction_score', 'Missing_Percent']:.1f}%)")
print(f"3. Duplicates found: {len(duplicates)} records")
print(f"4. Date formats are inconsistent (needs cleaning)")
print(f"5. Channel names have case issues (Email, email, EMAIL)")
print("\nNext: Tomorrow we'll clean all these issues!")
```
