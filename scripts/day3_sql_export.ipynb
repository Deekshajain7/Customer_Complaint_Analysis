```python
# Day 3: Export to SQL Database
# Goal: Load cleaned data into SQLite database

import pandas as pd
import sqlite3

print("="*60)
print("EXPORTING DATA TO SQL DATABASE")
print("="*60)

# Load enriched data
df = pd.read_csv('customer_complaints_enriched.csv')
print(f"\nLoaded {len(df)} rows from enriched dataset")

# Create SQLite database connection
database_name = 'complaints_analysis.db'
conn = sqlite3.connect(database_name)
print(f"Connected to database: {database_name}")

# Export main table to SQL
table_name = 'complaints'
df.to_sql(table_name, conn, if_exists='replace', index=False)
print(f"\n✅ Data exported to table: {table_name}")

# Verify export
query = "SELECT COUNT(*) as total_rows FROM complaints"
result = pd.read_sql_query(query, conn)
print(f"Verified: {result['total_rows'][0]} rows in database")

# Show table structure
query = "PRAGMA table_info(complaints)"
schema = pd.read_sql_query(query, conn)
print("\nTable schema:")
print(schema[['name', 'type']])

# Test query: Show first 5 complaints
print("\n" + "="*60)
print("TEST QUERY: First 5 complaints")
print("="*60)
query = """
SELECT complaint_id, date_submitted, complaint_category, 
       resolution_status, satisfaction_score
FROM complaints
LIMIT 5
"""
test_result = pd.read_sql_query(query, conn)
print(test_result)

# Close connection
conn.close()
print("\n✅ Database connection closed")
print(f"Database file: {database_name}")
```
